**Worum geht es?**
Der Verkehr in Städten nimmt zu. Das führt zu Staus, Zeitverlust, mehr Emissionen und höherem Druck auf die Infrastruktur. Ziel des Projekts ist eine zuverlässige Kurzfrist-Prognose: Wir wollen möglichst gut vorhersagen, wie hoch das Verkehrsaufkommen in den nächsten Stunden sein wird.


## Warum dieses Projekt wichtig ist

Heute wird Verkehr oft erst dann gesteuert, wenn das Problem bereits sichtbar ist. Mit belastbaren Prognosen kann früher reagiert werden, zum Beispiel bei:
- Ampelsteuerung
- Umleitungsmanagement
- Schicht- und Einsatzplanung in Leitstellen
- Verkehrsplanung für Liefer- und Pendlerzeiten

**Erwarteter Effekt:** weniger reaktive Entscheidungen, mehr vorausschauende Verkehrssteuerung.


## Zentrale Forschungsfrage

> **Wie gut lässt sich das kurzfristige Verkehrsaufkommen aus vergangenen Verkehrsdaten und äußeren Einflüssen (z. B. Wetter, Feiertage) vorhersagen, und welches Modell liefert im realistischen Betrieb die beste Qualität?**

### Unterfragen in einfacher Sprache
1. Gibt es erkennbare Muster nach Uhrzeit, Wochentag oder Saison?
2. Wie stark beeinflussen Wetter und Feiertage den Verkehr?
3. Bringen zusätzliche abgeleitete Merkmale (z. B. Mittelwerte, Werte aus den letzten Stunden) einen klaren Mehrwert?
4. Welcher Modelltyp ist am zuverlässigsten: einfache Verfahren, spezialisierte Zeitreihenmodelle oder Deep Learning?
5. Welches Modell ist der beste Kompromiss aus Genauigkeit, Rechenaufwand und Verständlichkeit?

## Datengrundlage

Wir arbeiten mit Zeitreihendaten in **stündlicher Auflösung**. Jede Zeile entspricht also einer Stunde.

### Wichtige Felder
- `ds`: Zeitpunkt (Datum + Uhrzeit)
- `y`: Verkehrsaufkommen (die Zielgröß, die wir vorhersagen)

### Zusätzliche Einflussfaktoren
- Stunde des Tages
- Wochentag
- Feiertags-Information
- Wetterlage (z. B. sonnig, bewölkt, Regen)
- optional weitere abgeleitete Größen (z. B. Durchschnittsgeschwindigkeit)

### Hinweis zur Datenquelle
- Für die Demonstration können realitätsnahe synthetische Daten genutzt werden.
- Reale CSV-Daten können später direkt eingebunden werden.


## Vorgehen im Projekt

1. **Problem und Zielbild festlegen**
   - Was genau soll vorhergesagt werden und in welchem Zeithorizont?
2. **Daten verstehen**
   - Muster, Ausreißer und Datenqualität prüfen
3. **Daten aufbereiten**
   - sinnvolle Zusatzmerkmale erzeugen, fehlende Werte behandeln
4. **Fair trainieren und testen**
   - zeitlich korrekt trennen (Vergangenheit trainiert, Zukunft testet)
5. **Modelle vergleichen**
   - mehrere Verfahren unter gleichen Bedingungen bewerten
6. **Ergebnisse verständlich aufbereiten**
   - Kennzahlen, Grafiken und klare Handlungsempfehlung


## Welche Modelle werden verglichen?

| Modell | Einfach erklärt | Vorteil | Zu beachten |
|---|---|---|---|
| **Lineare Regression** | Basisverfahren mit klaren Zusammenhängen | sehr verständlich und schnell | bei komplexen Mustern begrenzt |
| **Random Forest** | viele Entscheidungsbäume im Team | robust bei nichtlinearen Effekten | schwerer zu erklären als lineare Modelle |
| **Prophet** (optional) | speziell für Zeitreihen mit Saisonalität | gut für regelmäßige Muster | benötigt saubere Zeitstruktur |
| **LSTM** (optional) | neuronales Netz für zeitliche Abfolgen | kann komplexe Verläufe lernen | höherer Rechen- und Datenbedarf |

**Wichtig:** Nicht das komplexeste Modell gewinnt automatisch, sondern das Modell mit dem besten Gesamtpaket für den realen Einsatz.


## Wie wird Erfolg gemessen?

Wir nutzen drei zentrale Qualitätsmaße und erklären sie transparent:
- **MAE**: durchschnittlicher absoluter Fehler (leicht verständlich)
- **RMSE**: bestraft große Fehler stärker
- **R²**: zeigt, wie gut das Modell die Schwankungen im Verkehr erklärt

Zusätzlich betrachten wir die **Betriebssicht**:
- Stabilität bei wechselnden Bedingungen
- Laufzeit und Rechenaufwand
- Nachvollziehbarkeit für Fachanwender
